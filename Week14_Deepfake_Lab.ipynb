**Step 3 - Generate Your Deepfake**

```python
#-------------------------------- Python Code ---------------------------------------
import imageio
import numpy as np
import matplotlib.pyplot as plt
from skimage.transform import resize
from IPython.display import HTML
import warnings
warnings.filterwarnings("ignore")

# Load the necessary modules
from demo import load_checkpoints, make_animation
from skimage import img_as_ubyte

# Upload your source image and driving video
from google.colab import files
print("Upload source image:")
uploaded = files.upload()
source_image_path = list(uploaded.keys())[0]

print("Upload driving video:")
uploaded = files.upload()
driving_video_path = list(uploaded.keys())[0]

# Load files
source_image = imageio.imread(source_image_path)
reader = imageio.get_reader(driving_video_path)
driving_video = []
for frame in reader:
    driving_video.append(frame)

# Load model
generator, kp_detector = load_checkpoints(
    config_path='config/vox-256.yaml',
    checkpoint_path='checkpoints/vox-cpk.pth.tar'
)

# Generate deepfake
predictions = make_animation(
    source_image,
    driving_video,
    generator,
    kp_detector,
    relative=True
)

# Save output
imageio.mimsave('generated.mp4', [img_as_ubyte(frame) for frame in predictions], fps=30)
print("Deepfake generated: generated.mp4")
```
